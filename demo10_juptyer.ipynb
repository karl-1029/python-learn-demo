1 测试json

```python
import json

from datetime import datetime

# 原始Python数据
py_data = {
    "name": "张三",
    "age": 25,
    "is_vip": True,
    "hobbies": ["读书", "跑步"],
    "info": None,
    "register_time": datetime.now()  # datetime默认不能序列化，用来演示default
}
print("原始Python数据：", py_data)
```

复用前面的变量

```python
# json.dumps(obj) Python对象 → json字符串
json_str1 = json.dumps(py_data, default=str)
print("默认dumps：\n", json_str1)
print("-"*50)

# indent + ensure_ascii=False 美化+中文正常显示（最常用）
json_str2 = json.dumps(py_data, indent=2, ensure_ascii=False, default=str)
print("美化+中文：\n", json_str2)
print("-"*50)

# sort_keys=True 按键排序
json_str3 = json.dumps(py_data, indent=2, ensure_ascii=False, sort_keys=True, default=str)
print("key排序：\n", json_str3)
print("-"*50)

# separators 压缩json（接口传参常用，省流量）
json_str4 = json.dumps(py_data, separators=(',', ':'), default=str)
print("压缩版：", json_str4)
```
